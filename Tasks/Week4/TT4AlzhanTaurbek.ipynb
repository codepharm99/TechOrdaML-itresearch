```python
class Kitap:
    def __init__(self, atau, avtory):
        self.atau = atau        
        self.avtory = avtory    

    def kitap_turaly_malimet(self):
        print(f"Кітап: {self.atau}, Авторы: {self.avtory}")


# Нысан (объект) жасау
kitap1 = Kitap("Абай жолы", "Мұхтар Әуезов")

# Әдісті шақыру
kitap1.kitap_turaly_malimet()
```

```python
import numpy as np

arr_a = np.arange(1, 10)

arr_b = np.arange(1, 10).reshape(3, 3)

print("arr_a:", arr_a)
print("arr_a shape:", arr_a.shape)
print("arr_a dtype:", arr_a.dtype)

print("\narr_b:\n", arr_b)
print("arr_b shape:", arr_b.shape)
print("arr_b dtype:", arr_b.dtype)
```

```python
import numpy as np
vec1 = np.random.randint(1, 11, size=5)
vec2 = np.random.randint(1, 11, size=5)

print("vec1:", vec1)
print("vec2:", vec2)
sum_vec = vec1 + vec2
print("\nҚосу нәтижесі:", sum_vec)

square_vec1 = vec1 ** 2
print("vec1 квадраттары:", square_vec1)

mean_vec1 = vec1.mean()
print("vec1 орташа мәні:", mean_vec1)
max_vec1 = vec1.max()
print("vec1 ең үлкен мәні:", max_vec1)
```
